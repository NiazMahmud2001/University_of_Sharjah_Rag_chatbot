```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time
from urllib.parse import urljoin
```

```python
def get_clean_body_html(url):
    chrome_options = Options()
    chrome_options.add_argument("--headless")
    chrome_options.add_argument("--disable-gpu")
    chrome_options.add_argument("--no-sandbox")
    driver = webdriver.Chrome(options=chrome_options)
    
    try:
        driver.get(url)
        time.sleep(5) # waiting 5s to load the js
        driver.set_page_load_timeout(25)
        soup = BeautifulSoup(driver.page_source, "html.parser")
        driver.quit()
        
        
        for tag in soup(["script", "svg", "img", "link", "meta", "style", "textarea", "iframe", "input"]):
            tag.decompose()
            
            
        for tag in soup.find_all(True):
            if "style" in tag.attrs:
                del tag.attrs["style"]
            if "class" in tag.attrs: 
                del tag.attrs["class"]
            if "id" in tag.attrs: 
                del tag.attrs["id"]
            if "for" in tag.attrs: 
                del tag.attrs["for"]
            if "value" in tag.attrs: 
                del tag.attrs["value"]
            if "name" in tag.attrs: 
                del tag.attrs["name"]
            if "type" in tag.attrs: 
                del tag.attrs["type"]
            if "aria-label" in tag.attrs: 
                del tag.attrs["aria-label"]
            if "target" in tag.attrs: 
                del tag.attrs["target"]
                
            if tag.name in ["div", "li", "button", "span", "header"]:
                tag.unwrap()
                
            if tag.name == "a" and "href" in tag.attrs:
                href = tag.attrs["href"]
                # If href is a relative URL, prepend base URL
                if not href.startswith("http"):
                    tag.attrs["href"] = urljoin(url.strip(), href.strip())
                    
        body = soup.body
        return ' '.join(str(body).split())
    except Exception as e:
        return "No"
```

```python
with open("links.txt", 'r', encoding="utf-8") as f: 
    data = f.readlines()
    for index, line in enumerate(data): 
        link , name = line.split("|")
        
        webFetchData = get_clean_body_html(link)
        if webFetchData == "No": 
            continue
        
        with open("webOrganizedData.txt", "a", encoding="utf-8") as webPart: 
            webPart.write(f"{link.strip().replace("\n", "")} {name.strip().replace("\n", "")} ||PILLONIAZMJK|| {webFetchData}\n")
            print(f"{index}:{link}|{name}")
        
```
