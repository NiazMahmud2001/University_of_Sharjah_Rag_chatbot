```python
import gradio as gd  
import requests

class GradioUi(): 
    def __init__(self): 
        self.url = "" 
        self.chatbot = gd.Chatbot(
            elem_id="mychatbot",
            label="It's just Demo",
            placeholder="Hello! How can I assist you today?",
            show_copy_button=True,
            layout="bubble",
            render_markdown=True,
            allow_file_downloads= False,
            sanitize_html = True,
            height = "#mychatbot{height: 65vh}"
        )
        
    def respond(self, message, history):
        payload = {"query": message}
        res = requests.post(self.url, json=payload)
        try:
            return res.json().get("answer", "No answer received.")
        except Exception as e:
            return f"Error: {str(e)}"

 
        
    def launchBot(self): 
        chat = gd.ChatInterface(
            fn=self.respond,
            chatbot=self.chatbot,
            title="Demo Bot",
            theme="default",
            examples=[
                ["is UAE Society compulsory course in Computer Science course? "],
                ["How to register a course ?"],
                ["can you tell me, how many credit I have completed till now?"],
                ["What are the prerequisites before I can take the junior project?"],
                ["When can I register for my junior project?"],
                ["Can I take my internship before completing 66 credit hours?"],
                ["Can I take a course from another college as a free elective?"],
                ["What are mark distribution for Cloud Computing course?"],
            ],
            type="messages",
            save_history = False, 
            fill_height=True,
            stop_btn = True,
            autoscroll	=True,
            css="#mychatbot{height: 65vh}"
        )

        chat.launch(share=True, server_name="0.0.0.0", server_port=8760)

if __name__ == "__main__":
    GradioUi().launchBot()
```
