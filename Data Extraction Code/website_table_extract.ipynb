```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
def extract_text_and_tables(url, collegeName=""):
    
    response = requests.get(url) 
    if response.status_code != 200:
        return f"Error: Unable to fetch page, status code {response.status_code}"
    
    soup = BeautifulSoup(response.text, "html.parser")
    
    tables = []
    table_tags = soup.find_all("table")
    header = ""
    
    for table in table_tags:
        rows = []
        for row in table.find_all("tr"):
            cells = [cell.get_text(strip=True).replace("\n", "")  for cell in row.find_all(["th", "td"])]
            rows.append(",  ".join(cells))
        
        if rows:  # Ensure table is not empty
            tables.append({"metadata": collegeName, "Content":" || ".join(rows)})
        header = "" 

    return tables
```

```python

url = "https://www.sharjah.ac.ae/en/Academics/Degree/Undergraduate/Information-Technology-Multimedia"
website_name = "Department of Computer Science :Bachelor of Science in Information Technology Multimedia"
tables = extract_text_and_tables(url, website_name)

for i, table in enumerate(tables):
    print("Metadata:", table['metadata'], ':', table['Content'])
```
