```python
import pymupdf4llm as ppdfLLm
```

```python
ppdf_reader = ppdfLLm.LlamaMarkdownReader()
book = ppdf_reader.load_data("Student_Handbook.pdf")
```

```python
def split_text_into_chunks(text, max_words=200):
    words = text.split() 
    chunks = []
    
    for i in range(0, len(words), max_words):
        chunk = " ".join(words[i:i + max_words]) 
        chunks.append(chunk)
    
    return chunks


def processLine(text): 
    text = text.split("\n")
    text = "".join([x for x in text if x != ''])
    return text
```

```python
def processData(pdfName): 
    #current_section = {"metadata":pdf_Metadata, "title": "Introduction", "content": ""}
    sections = []
    
    for chunk in book: 
        data = processLine(chunk.text)
        if len(data) > 200: 
            chunks = split_text_into_chunks(data)
            for chunk in chunks: 
                if len(chunk)<15:
                    continue
                sections.append({"metadata":pdfName, "content": chunk})
        else:
            if len(data)<15:
                    continue
            sections.append({"metadata":pdfName, "content": data})
    return sections
```

```python
allData = processData("University of Sharjah Student Handbook") 

for x in allData: 
    print('metadata:',x["metadata"] , ' || content:', x["content"])
```
