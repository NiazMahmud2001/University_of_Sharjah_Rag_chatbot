```python
def text_formatter(text:str): 
    text = text.replace("\n", "")
    text = text.replace("  ", "")
    return text
```

```python
def open_and_read_pdf(path_pdf:str)->list[dict]: 
    doc = fitz.open(path_pdf)
    pages_and_chunks  = []
    
    for page_num, page in tqdm(enumerate(doc)):
        #text = page.get_text().encode('utf-8')
        text = page.get_text()
        cleanedText = text_formatter(text)
        sentences = nlp(cleanedText).sents
        cleaned_sentences = [str(sentence) for sentence in sentences]
        
        for sentence_chunk in cleaned_sentences:
            chunk_dict = {}
            chunk_dict["page_number"] = page_num-4
            
            joined_sentence_chunk = "".join(sentence_chunk).replace("  ", " ").strip()
            joined_sentence_chunk = re.sub(r'\.([A-Z])', r'. \1', joined_sentence_chunk)
            chunk_dict["sentence_chunk"] = joined_sentence_chunk
            
             # Get stats about the chunk
            chunk_dict["chunk_char_count"] = len(joined_sentence_chunk)
            chunk_dict["chunk_word_count"] = len([word for word in joined_sentence_chunk.split(" ")])
            chunk_dict["chunk_token_count"] = len(joined_sentence_chunk) / 4 # 1 token = ~4 characters
            
            pages_and_chunks.append(chunk_dict)
            
    return pages_and_chunks
```

```python
def retrieve_relevant_resources(query:str, embeddings:torch.tensor, model:SentenceTransformer=embedding_model, n_resources_to_return: int=5, print_time: bool=True):
    query_embedding = model.encode(query, convert_to_tensor=True)
    
    start_time = timer()
    dot_scores = util.dot_score(query_embedding, embeddings)[0]
    end_time = timer()

    if print_time:
        print(f"[INFO] Time taken to get scores on ({len(embeddings)} embeddings: {end_time-start_time:.5f} seconds.")

    scores, indices = torch.topk(input=dot_scores,k=n_resources_to_return)
    return scores, indices 
```

```python
def print_top_results_and_scores(query: str, embeddings: torch.tensor, pages_and_chunks: list[dict]=pages_and_chunks, n_resources_to_return: int=5):

    scores, indices = retrieve_relevant_resources(query=query, 
                                                  embeddings=embeddings, 
                                                  n_resources_to_return=n_resources_to_return,
                                                  print_time=False)

    for score, idx in zip(scores, indices):
        print(f"Score: {score:.4f}")
        print("Text:")
        print_wrapped(pages_and_chunks[idx]["sentence_chunk"])
        print(f"Page number: {pages_and_chunks[idx]['page_number']}")
        print("\n")
```

<h2><b>Define Groq => llama3-70B model</b></h2> 

```python
def summerizePrompt(basePrompt:str, temperature:float=1.0, max_completion_tokens:int=1024, top_p:float=1.0, stream:bool=True):
    client = Groq(api_key=api_key)
    completion = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "system",
                "content": "Rewrite the following passage in a more concise and structured manner, preserving all key details, you can rephrase, reformat and rewrite the text as needed also improve the clarity and conciseness also you can expand on the following response"
            },
            {
                "role": "user",
                "content": basePrompt
            }
        ],
        temperature=1,
        max_completion_tokens=max_completion_tokens,
        top_p=top_p,
        stream=stream,
        stop=None,
    )
    
    
    data = ""
    for chunk in completion:
        data += chunk.choices[0].delta.content or ""
    return data
```

<h2><b>Augmenting our prompt with context items with prompt engineering</b></h2> 

```python
def prompt_formatter(query: str, context_items: list[dict]) -> str:
        context = "- " + "\n- ".join([item["sentence_chunk"] for item in context_items])
        #print(context)
        base_prompt = """ 
        Context: {context} 
        User query: {query}
        Answer:
        """ 
        
        base_prompt = base_prompt.format(context=context, query=query)

        return base_prompt
```

```python
def ask(query: str, temperature: float=0.3, max_new_tokens:int=256, return_answer_only=True):
    scores, indices = retrieve_relevant_resources(query=query, embeddings=embeddings)
    # print("Scores: ", scores)
    context_items = [pages_and_chunks[i] for i in indices] 
    
    for i, item in enumerate(context_items): 
        item["score"] = scores[i].cpu() 

    # AUGMENTATION
    prompt = prompt_formatter(query=query, context_items=context_items)
    
    #GENERATION
    output_text = summerizePrompt(basePrompt=prompt, temperature=temperature, max_completion_tokens=max_new_tokens)

    if return_answer_only:
        return output_text

    return output_text, context_items
```

```python
class QueryRequest(BaseModel):
    query: str
    isChat: bool
```

```python
@app.get("/")
def home():
    return {"answer": "FastAPI is running inside Jupyter Notebook!"}

@app.post("/askQuestion")
@app.post("/askQuestion/")
def askQuestion(data: QueryRequest): 
    query = data.query
    answer = ask(query=query, temperature=1, return_answer_only=True)
    answer = answer.split("\n")
    breakLineAns = "" 
    for ans in answer:
        breakLineAns += ans + "<br>"
    return {"answer": breakLineAns}
```
